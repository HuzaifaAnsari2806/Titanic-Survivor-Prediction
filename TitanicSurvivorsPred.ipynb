```python
X=df.drop(columns=['survived'],axis=1)
y=df['survived']
```

```python
X
```

```python
from sklearn.preprocessing import LabelEncoder

encode=LabelEncoder()

columns=['sex','embarked','alone']

X_Label=X

for col in columns:
    X_Label[col]=encode.fit_transform(X_Label[col])
```

```python
X_Label
```

```python
X_Label.parch.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler

cols=X_Label.drop(['sex','alone','pclass','sibsp','parch','embarked'],axis=1).columns
scaler=StandardScaler()

X_Label[cols]=scaler.fit_transform(X_Label[cols])
```

```python
X_Label.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_Label, y, test_size=0.20, random_state=42)
```

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

model=LogisticRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score

confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
print("Accuracy: ", accuracy_score(y_test, y_pred))
```

# KNN Algorithm

```python
from sklearn.neighbors import KNeighborsClassifier

model= KNeighborsClassifier(n_neighbors=43)
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score

confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
print("Accuracy: ", accuracy_score(y_test, y_pred))
```

# Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB

model=GaussianNB()
y_pred=model.fit(X_train,y_train).predict(X_test)
```

```python
y_pred
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```
